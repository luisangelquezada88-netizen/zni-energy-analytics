# Lectura y preprocesamiento (data wrangling)

## Estado de la prestación del servicio

```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/servicio_electrico.csv")
df.head()
```

```python
df.info()
```

```python
df = df.rename(columns={
    'ID DEPATAMENTO': 'id_departamento',  # ojo: viene con typo DEPATAMENTO
    'DEPARTAMENTO': 'departamento',
    'ID MUNICIPIO': 'id_municipio',
    'MUNICIPIO': 'municipio',
    'ID LOCALIDAD': 'id_localidad',
    'LOCALIDAD': 'localidad',
    'AÑO SERVICIO': 'anio_servicio',
    'MES SERVICIO': 'mes_servicio',
    'ENERGÍA ACTIVA': 'energia_activa',
    'ENERGÍA REACTIVA': 'energia_reactiva',
    'POTENCIA MÁXIMA': 'potencia_maxima',
    'DÍA DE DEMANDA MÁXIMA': 'dia_demanda_max',
    'FECHA DE DEMANDA MÁXIMA': 'fecha_demanda_max',
    'PROMEDIO DIARIO EN HORAS': 'promedio_horas'
})

df.head()
```

```python
# Energia y potencia: el punto es miles -> se quita; la coma es decimal -> punto
cols_miles = ['energia_activa', 'energia_reactiva', 'potencia_maxima']

for col in cols_miles:
    df[col] = df[col].astype(str).str.replace('.', '', regex=False).str.replace(',', '.', regex=False)
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Horas 0-24: la coma es decimal; el punto YA es decimal ingles (2.400 = 2.4h) -> no quitarlo
df['promedio_horas'] = df['promedio_horas'].astype(str).str.replace(',', '.', regex=False)
df['promedio_horas'] = pd.to_numeric(df['promedio_horas'], errors='coerce')
df['promedio_horas'] = df['promedio_horas'].clip(upper=24)

cols_num = cols_miles + ['promedio_horas']
df[cols_num].dtypes
```

```python
df[cols_num].describe()
```

```python
df['fecha_demanda_max'] = pd.to_datetime(df['fecha_demanda_max'], format='mixed', dayfirst=True, errors='coerce')
df.isna().sum()
```

```python
# Texto: quitar espacios y pasar a minuscula para evitar duplicados (LETICIA vs Leticia)
cols_txt = ['departamento', 'municipio', 'localidad', 'dia_demanda_max']

for col in cols_txt:
    df[col] = df[col].astype(str).str.strip().str.lower()

df[cols_txt].head()
```

```python
# IDs: el shape trae texto con cero a la izquierda ('05'); el CSV trae entero (5). Se igualan a texto.
# 1 fila trae id 0 siendo CAUCA (mpio 19809): se corrige a 19
df['id_departamento'] = df['id_departamento'].replace(0, 19)

# DANE: depto 2 digitos, mpio 5, localidad 8. zfill conserva el cero inicial.
df['id_departamento'] = df['id_departamento'].astype(str).str.zfill(2)
df['id_municipio'] = df['id_municipio'].astype(str).str.zfill(5)
df['id_localidad'] = df['id_localidad'].astype(str).str.zfill(8)

df[['id_departamento', 'id_municipio', 'id_localidad']].dtypes
```

## Decisiones tecnicas del wrangling

**1. Lectura (`info`).** 6216 filas, 14 columnas. 5 `int64` bien; 9 `str` donde 4 debian ser numero y 1 fecha. Formato colombiano: miles con punto (`3.930.642`) y decimal con coma (`7.768,76`).

**2. Nombres (`rename`).** Minuscula `snake_case`, sin tildes ni `Ñ` (`anio_servicio`, `energia_activa`). Se corrige el typo de origen `ID DEPATAMENTO` -> `id_departamento`.

**3. Numeros.** Energia y potencia: el punto es miles -> se quita; la coma es decimal -> punto (`to_numeric`). Horas 0-24: la coma es decimal pero el punto YA es decimal ingles (`2.400` = 2.4 h) -> no quitarlo; se topa con `clip(upper=24)`. Sin esto el `describe` daba `max 2400`.

**4. Fecha (`to_datetime`).** `format='mixed', dayfirst=True` para el formato `2020 Jan 27 02:15:00 PM` sin `UserWarning`.

**5. Texto.** `strip().lower()` en departamento/municipio/localidad/dia para evitar duplicados (`LETICIA` vs `Leticia`). Mayuscula solo al unir con el shape del DANE.

**6. Nulos (`isna`).** 84 en potencia/dia/fecha son las mismas filas: Choco y Buenaventura, ene-2025, con energia y horas en 0 (sin servicio no hay pico que medir). Se exportan como `NaN`; el clustering filtra con `promedio_horas > 0`.
